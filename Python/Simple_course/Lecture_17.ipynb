```python
class Student:
    def __init__(self, name, grade, num_inter):
        self.name = name
        self.grade = grade
        self.num_inter = num_inter
    def __str__(self):
      return "the name is "+ self.name +  "the grade is "+ str(self.grade)

    def __eq__(self, other):
      if (self.grade == other.grade) and (self.num_inter == other.num_inter):
        return True
      return False

s1 = Student("bob", 110, 3)
s2 = Student("alice", 110, 3)
print(s1)

print(s1 == s2)
```

```python
class Quarterback:
    def __init__(self, yrds, tds, cmps, atts, ints, wins):
        self.wins = wins

        # Calculate quarterback passer rating (NCAA)
        self.rating = ((8.4 * yrds) + (330 * tds) + (100 * cmps) -
                       (200 * ints)) / atts

    def __lt__(self, other):
        if (self.rating < other.rating) or (self.wins < other.wins):
            return True
        return False

    def __gt__(self, other):
      if (self.rating > other.rating) and (self.wins > other.wins):
        return True
      return False


peyton = Quarterback(yrds=4700, atts=679, cmps=450, tds=33, ints=17, wins=10)
eli = Quarterback(yrds=4002, atts=539, cmps=339, tds=31, ints=25, wins=9)

if peyton > eli: ## Calles __lgt__
    print("Peyton is the better QB")
elif peyton < eli: # calles __lt__
    print("Eli is the better QB")
else:
    print("It is not clear who the better QB is...")
```

```python
class Time24:
    def __init__(self, hours, minutes):
        self.hours = hours
        self.minutes = minutes

    def __str__(self):
        return f"{self.hours:02d}:{self.minutes:02d}"

    def __gt__(self, other):
        if self.hours > other.hours:
            return True
        else:
            if self.hours == other.hours:
                if self.minutes > other.minutes:
                    return True
        return False

    def __sub__(self, other):
        """ Calculate absolute distance between two times """
        if self > other:
            larger = self
            smaller = other
        else:
            larger = other
            smaller = self

        hrs = larger.hours - smaller.hours
        mins = larger.minutes - smaller.minutes
        if mins < 0:
            mins += 60
            hrs -=1

        # Check if times wrap to new day
        if hrs > 12:
            hrs = 24 - (hrs + 1)
            mins = 60 - mins
            if mins >= 60:
                mins -= 60
                hrs += 1

        # Return new Time24 instance
        return Time24(hrs, mins)

t1 = input("Enter time1 (hours:minutes): ")
tokens = t1.split(":")
time1 = Time24(int(tokens[0]), int(tokens[1]))

t2 = input("Enter time2 (hours:minutes): ")
tokens = t2.split(":")
time2 = Time24(int(tokens[0]), int(tokens[1]))

print(f"Time difference: {time1 - time2}")
```

```python
class BankAccount:
    def __init__(self, account_holder, initial_balance=0):
        self.account_holder = account_holder
        self.balance = initial_balance
        print(f"Account created for {self.account_holder} with initial balance ${self.balance:.2f}")

    def _log_transaction(self, transaction_type, amount): # A 'private' internal helper method
        print(f"  [Log] {transaction_type} of ${amount:.2f} processed.")

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            self._log_transaction("Deposit", amount)
            print(f"Deposited ${amount:.2f}. New balance: ${self.balance:.2f}")
        else:
            print("Deposit amount must be positive.")

    def withdraw(self, amount):
        if amount > 0:
            if self.balance >= amount:
                self.balance -= amount
                self._log_transaction("Withdrawal", amount)
                print(f"Withdrew ${amount:.2f}. New balance: ${self.balance:.2f}")
            else:
                print(f"Insufficient funds. Current balance: ${self.balance:.2f}")
        else:
            print("Withdrawal amount must be positive.")

    def get_balance(self):
        print(f"Current balance for {self.account_holder}: ${self.balance:.2f}")
        return self.balance


# --- Demonstrating the BankAccount class ---
print("\n--- Creating and using a BankAccount object ---")
my_account = BankAccount("Alice Wonderland", 500.00)

my_account.get_balance()
my_account.deposit(200.00)
my_account.withdraw(150.00)
my_account.withdraw(600.00) # This should fail due to insufficient funds
my_account.get_balance()

```

```python
import math

class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __str__(self):
        return f"Vector({self.x}, {self.y})"

    # Customizing comparison operators
    def __lt__(self, other): # Less than: v1 < v2 # v1= self, v2 = other   v1 < v2
        # Compare by magnitude
        return math.sqrt(self.x**2 + self.y**2) < math.sqrt(other.x**2 + other.y**2)

    def __ge__(self, other): # Greater than or equal to: v1 >= v2
        # Compare by magnitude
        return math.sqrt(self.x**2 + self.y**2) >= math.sqrt(other.x**2 + other.y**2)

    # Customizing arithmetic operators
    def __add__(self, other): # Addition: v1 + v2
        if isinstance(other, Vector): # checking if  other are vector
            return Vector(self.x + other.x, self.y + other.y)
        else:
            raise TypeError("Can only add a Vector to another Vector")

    def __sub__(self, other): # Subtraction: v1 - v2
        if isinstance(other, Vector):
            return Vector(self.x - other.x, self.y - other.y)
        else:
            raise TypeError("Can only subtract a Vector from another Vector")

    def __mul__(self, scalar): # Multiplication (scalar): v1 * 2
        if isinstance(scalar, (int, float)):
            return Vector(self.x * scalar, self.y * scalar)
        else:
            raise TypeError("Can only multiply a Vector by a scalar (int or float)")

# --- Demonstration ---
v1 = Vector(3, 4)
v2 = Vector(1, 2)
v3 = Vector(6, 8)

print(f"v1: {v1}")
print(f"v2: {v2}")
print(f"v3: {v3}")

# Comparison operations
print(f"v1 < v2: {v1 < v2}") # Expected: False (magnitude of v1 is 5, v2 is ~2.24)
print(f"v1 >= v2: {v1 >= v2}") # Expected: True
print(f"v1 < v3: {v1 < v3}") # Expected: True (magnitude of v3 is 10)

# Arithmetic operations
v_sum = v1 + v2
print(f"v1 + v2: {v_sum}") # Expected: Vector(4, 6)

v_diff = v1 - v2
print(f"v1 - v2: {v_diff}") # Expected: Vector(2, 2)

v_scaled = v1 * 2
print(f"v1 * 2: {v_scaled}") # Expected: Vector(6, 8)

v_scaled_float = v2 * 0.5
print(f"v2 * 0.5: {v_scaled_float}") # Expected: Vector(0.5, 1.0)
```

very important table

| Method | Description |
| :----------------- | :-------------------------- |
| `__add__(self, other)` | Add (+) |
| `__sub__(self, other)` | Subtract (-) |
| `__mul__(self, other)` | Multiply (*) |
| `__truediv__(self, other)` | Divide (/) |
| `__floordiv__(self, other)` | Floored division (//) |
| `__mod__(self, other)` | Modulus (%) |
| `__pow__(self, other)` | Exponentiation (**) |
| `__and__(self, other)` | "and" logical operator |
| `__or__(self, other)` | "or" logical operator |
| `__abs__(self)` | Absolute value (abs()) |
| `__int__(self)` | Convert to integer (int()) |
| `__float__(self)` | Convert to floating point (float()) |
