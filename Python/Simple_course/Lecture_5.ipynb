```python
my_list = ["22", 44, 55," bob"]
print(my_list)
# append
my_list.append(100)
print(f"my list: {my_list}")
# remove
my_list.remove(44) # remove by value
print(f"my list: {my_list}")
# pop
my_list.pop(0) # remove by index
print(f"my list: {my_list}")
```

```python
# methods for list
my_list = [22,33,445,55,22,22]
#len
print(f"length of list: {len(my_list)}")
# l + l2
print(f"my list: {my_list}")
my_list2 = [11,22,33]
print(f"my list2: {my_list2}")
my_list3 = my_list + my_list2
print(f"my list3: {my_list3}")
# max
print(f"max of list: {max(my_list)}")
# sum
print(f"sum of list: {sum(my_list)}")
# we find value 22
print(f"index of 22 is: {my_list.index(22)}")
# count 22
print(f"count of 22 is: {my_list.count(22)}")
```

Name type using tuple

```python
from collections import namedtuple

Car = namedtuple("Car", ["make","model","price","horsepower","seats"])  # Create the named tuple

chevy_blazer = Car("Chevrolet", "Blazer", 32000, 275, 8)  # Use the named tuple to describe a car
chevy_impala = Car("Chevrolet", "Impala", 37495, 305, 5)  # Use the named tuple to describe a different car

print(chevy_blazer)
print(chevy_impala)

# access information
print("the make is ", chevy_blazer.make)
print("the price is ", chevy_blazer.price)
```

```python
# Program to calculate statistics from student test scores.
midterm_scores = [99.5, 78.25, 76, 58.5, 100, 87.5, 91, 68, 100]
final_scores = [55, 62, 100, 98.75, 80, 76.5, 85.25]

# Combine the scores into a single list
all_scores = midterm_scores + final_scores

num_midterm_scores = len(midterm_scores)
num_final_scores = len(final_scores)

print(f"{num_midterm_scores} students took the midterm.")
print(f"{num_final_scores} students took the final.")

# Calculate the number of students who took the midterm but not the final
dropped_students = num_midterm_scores - num_final_scores
print(f"{dropped_students} students must have dropped the class.")

# Your code goes here
# for midterm_scores
min_mid = min(midterm_scores)
max_min = max(midterm_scores)
# the range
print(f"the range of midterm scores is {max_min - min_mid} and the values are between {min_mid} and {max_min}")
# cal the avg  = sum / len
avg_mid = sum(midterm_scores) / len(midterm_scores)
print(f"the avg of midterm scores is {avg_mid:.2f}")
```
