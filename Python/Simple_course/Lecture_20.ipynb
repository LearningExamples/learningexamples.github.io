```python
# Demonstrating 'write()' with 'w' mode (write, overwrites file if it exists)
print("Demonstrating 'write()' with 'w' mode and 'a' mode:")
with open('write_example.txt', 'w') as f:
    f.write('hello 5+7 =12 ?\n')
    f.write('This is a new line.\n')
    f.write('This is another line.') # \n on the last line would create an empty line


```

| Mode | Description | Allow read? | Allow write? | Create missing file? | Overwrite file? |
|---|---|---|---|---|---|
| "r" | Open the file for reading. | Yes | No | No | No |
| "w" | Open the file for writing. If the file does not exist, then the file is created. Contents of an existing file are overwritten. | No | Yes | Yes | Yes |
| "a" | Open the file for appending. If the file does not exist, then the file is created. Writes are added to the end of existing file contents. | No | Yes | Yes | No |

```python
import time

file_name = 'flush_example.txt'

print(f"Opening '{file_name}' for writing...")
with open(file_name, 'w') as f:
    f.write('This is the first line. It will be flushed.')
    print("Wrote 'This is the first line. It will be flushed.' to buffer. Now flushing...")
    f.flush() # Force writing buffered data to disk
    print("Buffer flushed. Check the file content now. (Pausing for 3 seconds)")
    time.sleep(3) # Pause to allow manual inspection of the file

    f.write('This is the second line. It will be written after the flush.\n')
    print("Wrote 'This is the second line. It will be written after the flush.' to buffer.")

print(f"File '{file_name}' operations complete. Content should now be fully written.")

# Optional: Read the content back to verify
print(f"\nReading content from '{file_name}':")
with open(file_name, 'r') as f_read:
    print(f_read.read())
```

```python
import os
import datetime

curr_day = datetime.date(1997, 8, 29)

num_days = 30
for i in range(num_days):
    year = str(curr_day.year)
    month = str(curr_day.month)
    day = str(curr_day.day)

    # Build path string using current OS path separator
    file_path = os.path.join("logs", year, month, day, "log.txt")
    print(f"File path: {file_path}")

    # Create directories if they don't exist
    os.makedirs(os.path.dirname(file_path),)

    # For this example, we'll create the file if it doesn't exist to avoid a FileNotFoundError when opening for reading.
    # In a real scenario, you might create the log content and then write it.
    with open(file_path, "w") as f_write:
        f_write.write(f"This is a log entry for {curr_day}\n")

    f = open(file_path, "r")

    print(f"{file_path}: {f.read()}")
    f.close()

    curr_day = curr_day + datetime.timedelta(days=1)
```

```python
# this is beyond the class level for now
import struct

# Define a format string:
# '<' indicates little-endian byte order
# 'i' for a standard integer (4 bytes)
# 'f' for a float (4 bytes)
# '?' for a boolean (1 byte)
# '10s' for a string of 10 characters (bytes)
format_string = '<if?10s' # Corrected: removed one 'f' to match provided variables

# Values to pack
my_int = 12345
my_float = 3.14159
my_bool = True
my_string = b"hello"

# Pack the values into a bytes object
packed_data = struct.pack(format_string, my_int, my_float, my_bool, my_string)
print(f"Original values: int={my_int}, float={my_float}, bool={my_bool}, string={my_string.decode()}")
print(f"Packed data (bytes): {packed_data}")
print(f"Size of packed data: {len(packed_data)} bytes\n")

# Unpack the bytes object back into Python values
unpacked_values = struct.unpack(format_string, packed_data)

# Access the unpacked values
unpacked_int = unpacked_values[0]
unpacked_float = unpacked_values[1]
unpacked_bool = unpacked_values[2]
unpacked_string = unpacked_values[3].decode().strip('\x00') # Decode and remove null bytes

print(f"Unpacked values: int={unpacked_int}, float={unpacked_float}, bool={unpacked_bool}, string='{unpacked_string}'")

# Verify if original and unpacked values match
print(f"\nVerification:")
print(f"Int match: {my_int == unpacked_int}")
print(f"Float match: {abs(my_float - unpacked_float) < 1e-6}") # Use tolerance for floats
print(f"Bool match: {my_bool == unpacked_bool}")
print(f"String match: {my_string.decode().strip() == unpacked_string.strip()}")
```

## CSV

```python
import csv
import random

# Define the CSV file name
csv_file_name = 'random_data.csv'

# Define the header for the CSV file
headers = ['ID', 'Name', 'Value', 'Status']

# Generate some random data
data = []
for i in range(1, 6):
    name = f"Item_{i}"
    value = round(random.uniform(10.0, 100.0), 2) # Random float between 10 and 100
    status = random.choice(['Active', 'Inactive', 'Pending'])
    data.append([i, name, value, status])

print(f"Generated random data:\n{data}\n")


```

```python
num_rows = len(data)
if num_rows > 0:
    num_cols = len(data[0])
    print(f"Data shape: {num_rows} rows x {num_cols} columns")
else:
    print("Data is empty.")
```

```python
display(data)
```

```python
#  Write the data to a CSV file
print(f"Writing data to '{csv_file_name}'...")


# csvfile = (csv_file_name, 'w', newline='') -- then need to close
with open(csv_file_name, 'w', newline='') as csvfile:
    # Create a CSV writer object
    csv_writer = csv.writer(csvfile)

    # Write the header row
    csv_writer.writerow(headers)

    # Write the data rows
    csv_writer.writerows(data) # you could do this in a loop as well same as the text book
    # or you can do it in a loop
    # for row in data:
    #     csv_writer.writerow(row)
print(f"Successfully wrote data to '{csv_file_name}'.")

```

```python
csvfile = open(csv_file_name, 'r', newline='')
csv_reader = csv.reader(csvfile)
for row in csv_reader:
    print(row)
    print("-"*50)

csvfile.close()
```

```python
# Optional: Read the CSV file back to verify
print(f"\nReading content from '{csv_file_name}' to verify:")
with open(csv_file_name, 'r', newline='') as csvfile:
    csv_reader = csv.reader(csvfile)
    for row in csv_reader:
        print(row)
        print("-"*50)
```
