```python
# Get student name from user input
student_name = "Bob Adams"

# Check if the object student_name is the same as the object "Amy Adams"
# This uses the identity operator 'is'
if student_name is "Bob Adams":
    print("Identity operator: True")
else:
    print("Identity operator: False")

# Check if the value of student_name is equal to the value "Amy Adams"
# This uses the equality operator '=='
if student_name == "Bob Adams":
    print("Equality operator: True")
else:
    print("Equality operator: False")
```

This code snippet illustrates the difference between the `is` and `==` operators in Python:

*   The `is` operator checks if two variables refer to the same object in memory (identity).
*   The `==` operator checks if the values of two variables are equal (equality).

In the example, when you enter "Amy Adams", the `==` operator will return `True` because the values are equal. However, the `is` operator will likely return `False` because the string you enter from the input is a new object in memory, even if its value is the same as the literal string `"Amy Adams"`.

##  Splitting and joining strings

```python
# Example string with spaces
sentence = "This is a sample sentence"
words = sentence.split()
print(f"Splitting '{sentence}' by default (whitespace): {words}")

# Example string with a different delimiter
data = "apple ,banana ,cherry"
fruits = data.split(',')
print(f"Splitting '{data}' by comma: {fruits}")

# Example with a specified number of splits
log_line = "ERROR: File not found: myfile.txt"
parts = log_line.split(':', 1)
print(f"Splitting '{log_line}' by colon with max split of 1: {parts}")
```

```python
# Example list of strings
words = ["This", "is", "a", "list", "of", "words"]

# Join with a space
sentence = " ".join(words)
print(f"Joining with a space: {sentence}")

# Join with a comma and space
csv_data = ", ".join(words)
print(f"Joining with comma and space: {csv_data}")

# Join with an underscore
underscore_string = "_".join(words)
print(f"Joining with an underscore: {underscore_string}")

# Example tuple of strings
data_tuple = ("user1", "user2", "user3")
joined_tuple = "-".join(data_tuple)
print(f"Joining a tuple with a hyphen: {joined_tuple}")
```

```python
# Example string with a specific separator
data_string = "item1;item2;item3;item4"

# Split the string by the semicolon
items = data_string.split(';')
print(f"After splitting by semicolon: {items}")

# Join the list of items with a different separator (e.g., a comma and space)
new_data_string = ", ".join(items)
print(f"After joining with comma and space: {new_data_string}")

# Example to replace spaces with underscores
sentence_with_spaces = "This is a sentence with spaces"
words_list = sentence_with_spaces.split() # Split by whitespace
sentence_with_underscores = "_".join(words_list) # Join with underscore
print(f"Replacing spaces with underscores: {sentence_with_underscores}")
```

```python
# Example string with a specific separator
data_string = "item1;item2;item3;item4"

# Replace the semicolon with a comma and space
new_data_string_replace = data_string.replace(';', ', ')
print(f"After replacing semicolon with comma and space: {new_data_string_replace}")

# Example to replace spaces with underscores using replace()
sentence_with_spaces = "This is a sentence with spaces"
sentence_with_underscores_replace = sentence_with_spaces.replace(' ', '_')
print(f"Replacing spaces with underscores using replace(): {sentence_with_underscores_replace}")
```
