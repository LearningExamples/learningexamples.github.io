```python
import numpy as np # as np is standard

# 1. Create two Python lists
list1 = [1, 2, 3]
list2 = [4, 5, 6]

print("Original Lists:")
print(f"List 1: {list1}")
print(f"List 2: {list2}")
print("\n" + "-" * 30 + "\n")

# 2. Convert lists to NumPy arrays
array1 = np.array(list1)
array2 = np.array(list2)

print("NumPy Arrays (converted from lists):")
print(f"Array 1: {array1}")
print(f"Array 2: {array2}")
```

```python

# 3. Perform addition and multiplication on lists
print("Operations on Lists:")
list_addition = list1 + list2
print(f"List Addition (list1 + list2): {list_addition} (concatenates lists)")

# 4. Perform addition and multiplication on NumPy arrays
print("Operations on NumPy Arrays:")
array_addition = array1 + array2
print(f"Array Addition (array1 + array2): {array_addition} (element-wise addition)")
```

```python

# --- 2. drop_duplicates() method ---
print("--- drop_duplicates() Examples ---")
# Create a dummy DataFrame with some duplicates for demonstration
data_with_dups = {
    'ID': [1, 2, 3, 1, 4],
    'Name': ['Alice', 'Bob', 'Charlie', 'Alice', 'David'],
    'Score': [85, 90, 78, 85, 92]
}
df_dups = pd.DataFrame(data_with_dups)
print("Original DataFrame with duplicates:")
print(df_dups)

df_no_dups = df_dups.drop_duplicates(subset=['Name', 'Score'])
print("\nDataFrame after dropping duplicates based on 'Name' and 'Score':")
print(df_no_dups)
print("\n" + "-" * 30 + "\n")
```

```python
# Load matplotlib.pyplot, pandas, and numpy
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# 1. Generate data using NumPy
x = np.linspace(0, 10, 100) # 100 points between 0 and 10
y = np.sin(x) + np.random.normal(0, 0.1, 100) # Sine wave with some noise

# 2. Put data into a Pandas DataFrame
data = {
    'X_values': x,
    'Y_values': y
}
df_plot = pd.DataFrame(data)

print("First 5 rows of the DataFrame used for plotting:")
print(df_plot.head())
print("\n" + "-" * 30 + "\n")

# 3. Plot the data using Matplotlib (default size)
plt.figure(figsize=(4, 2)) # Default or desired size for the first plot
plt.plot(df_plot['X_values'], df_plot['Y_values'], label='Sine wave with noise', color='skyblue')
plt.title('Original Plot (Default Size)')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True)
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Reuse df_plot from previous steps, or create new if not available
# If df_plot doesn't exist, uncomment and run these lines:
x = np.linspace(0, 10, 50) # Fewer points for clearer scatter
y = np.sin(x) + np.random.normal(0, 0.2, 50) # Sine wave with some noise
df_plot = pd.DataFrame({'X_values': x, 'Y_values': y})

plt.figure(figsize=(10, 6))

# Create a scatter plot with stars as markers
plt.scatter(df_plot['X_values'], df_plot['Y_values'], marker='>', s=100, color='purple', label='Data Points (Stars)')

# Use title()
plt.title('Scatter Plot of Sine Wave with Noise')

# Use xlabel() and ylabel()
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')

# Use text(x, y, s)
plt.text(1, 0.8, 'Important Area', fontsize=12, color='green', bbox=dict(facecolor='white', alpha=0.5))

# Use annotate(s, xy, xytext)
# Let's annotate the highest point in the displayed data
max_y_index = df_plot['Y_values'].idxmax()
max_x = df_plot['X_values'][max_y_index]
max_y = df_plot['Y_values'][max_y_index]
plt.annotate(
    f'Peak: ({max_x:.2f}, {max_y:.2f})',
    xy=(max_x, max_y),
    xytext=(max_x + 1.5, max_y + 0.5),
    arrowprops=dict(facecolor='black', shrink=0.05),
    fontsize=10,
    color='blue'
)

# Use legend(loc)
plt.legend(loc='upper right')

plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Ensure df_plot exists for sine wave data
# If df_plot doesn't exist, create it (e.g., from previous steps)
if 'df_plot' not in locals():
    x = np.linspace(0, 10, 100)
    y = np.sin(x) + np.random.normal(0, 0.1, 100)
    df_plot = pd.DataFrame({'X_values': x, 'Y_values': y})

# Generate data for a second plot (cosine wave)
x_cosine = df_plot['X_values']
y_cosine = np.cos(x_cosine) + np.random.normal(0, 0.1, len(x_cosine)) # Corrected length to match x_cosine

# Create a figure and a set of subplots
plt.figure(figsize=(12, 5))

# Subplot 1: Sine wave
plt.subplot(1, 2, 1) # (rows, columns, plot_number)
plt.plot(df_plot['X_values'], df_plot['Y_values'], color='skyblue', label='Sine Wave with Noise')
plt.title('Sine Wave')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True)
plt.legend()

# Subplot 2: Cosine wave
plt.subplot(1, 2, 2) # (rows, columns, plot_number)
plt.plot(x_cosine, y_cosine, color='lightcoral', label='Cosine Wave with Noise')
plt.title('Cosine Wave')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True)
plt.legend()

plt.tight_layout() # Adjust subplot parameters for a tight layout
plt.show()
```
