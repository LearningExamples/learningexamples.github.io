```python
i = 0
j = 0
while i < len(fruits):
    # print(fruits[i])
    j = 0
    while j < len(fruits[i]):
        # print(fruits[i][j])
        print(f"i: {i}, j: {j}")
        j += 1
    i += 1
```

```python
# Example 1: While loop outside, for loop inside
print("Example 1: While loop outside, for loop inside")
i = 0
while i < len(fruits):
    print(f"Outer loop (while) iteration: {i}")
    for fruit in fruits[i]:
        print(f"\tInner loop (for) fruit: {fruit}")
    i += 1

print("\n" + "="*20 + "\n")

# Example 2: For loop outside, while loop inside
print("Example 2: For loop outside, while loop inside")
for fruit_list in fruits:
    print(f"Outer loop (for) processing list: {fruit_list}")
    j = 0
    while j < len(fruit_list):
        print(f"\tInner loop (while) fruit: {fruit_list[j]}")
        j += 1
```

# Break and continue

```python
# Example with break
print("Example with break:")
for i in range(10):
    if i == 5:
        break  # Exit the loop when i is 5
    print(i)

print( "="*20 + "\n")

```

```python
# Example with continue
print("Example with continue:")
for i in range(5):
    print("+++++++++++++++ before my if i=",i)
    if i == 2:
        continue  # Skip the rest of the code in this iteration when i is 5
    print(" after my if i=",i)
```

```python
# Example with break in a while loop
print("Example with break in a while loop:")
i = 0
while i < 10:
    if i == 5:
        break  # Exit the loop when i is 5
    print(i)
    i += 1

print("\n" + "="*20 + "\n")

```

```python
# Example with continue in a while loop
print("Example with continue in a while loop:")
i = 0
while i < 5:
    print("in the looop and i= ",i)
    if i == 2:
        i += 1 # Increment before continue to avoid infinite loop
        continue  # Skip the rest of the code in this iteration when i is 5
    print(i)
    i += 1
```

```python
# Example with break in a nested loop
print("Example with break in a nested loop:")
for i in range(3):
    if i == 1:
        break  # Exit the outer loop when i is 1
    for j in range(3):
        print(f"i: {i}, j: {j}")
    print("End of inner loop for i =", i)
print("\n" + "="*20 + "\n")

```

```python
# Example with break in a nested loop
print("Example with break in a nested loop:")
for i in range(3):
    for j in range(3):
        if i == 1 and j == 1:
            break  # This breaks out of the inner loop only
        print(f"i: {i}, j: {j}")
    print("End of inner loop for i =", i)
print("\n" + "="*20 + "\n")

```

```python
# Example with continue in a nested loop
print("Example with continue in a nested loop:")
for i in range(3):
    for j in range(3):
        if i == 1 and j == 1:
            continue  # This skips the rest of the inner loop for this iteration
        print(f"i: {i}, j: {j}")
    print("End of inner loop for i =", i)
```

```python
# Example with continue in a nested loop
print("Example with continue in a nested loop:")
for i in range(3):
    if i == 1:
        continue  # Skip the rest of the inner loop for this iteration
    for j in range(3):
        print(f"i: {i}, j: {j}")
    print("End of inner loop for i =", i)
```

```python
# Example with while loop and else block
print("Example with while loop and else block:")
i = 0
while i < 5:
    print(i)
    i += 1
else:
    print("Loop finished successfully (condition became false)")
```

```python
# Example with for loop and else block
print("Example with for loop and else block:")
for i in range(5):
    print(i)
else:
    print("Loop finished successfully (iterated through the sequence)")
```

```python
# Example with enumerate
print("Example with enumerate:")
my_list = ["a", "b", "c"]
for index, value in enumerate(my_list):
    print(f"Index: {index}, Value: {value}")

print("\n" + "="*20 + "\n")



my_list_of_tuples = [(1, 'one'), (2, 'two')]
for number, word in my_list_of_tuples:
    print(f"Number: {number}, Word: {word}")
```

```python
# Example with unpacking
print("Example with unpacking:")
my_tuple = (1, 2, 3)
a = my_tuple[0]
b = my_tuple[1]
c = my_tuple[2]
print(f"a: {a}, b: {b}, c: {c}")

print("\n" + "="*20 + "\n")
a, b, c = my_tuple
print(f"a: {a}, b: {b}, c: {c}")
```
