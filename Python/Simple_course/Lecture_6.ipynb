```python
# sets
first_names = ["Alba", "Hema", "Ron", "Alba", "Musa", "Ron", "Ron"]
print("my list is :", first_names)
# Creating a set removes any duplicate values
names_set = set(first_names)

print("my set is: ",names_set)

# demonstrate set operations
print("Length of the set:", len(names_set))

set2 = {"David", "Sarah", "Ron"}
print("Set 2:", set2)
names_set.update(set2) # this UPDATES the name_set not create a new set of both!!!
print("Set after update:", names_set)
```

```python
# set operations
set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8}
set3 = {5, 6, 9, 10}

# Intersection
print("Intersection of set1 and set2:", set1.intersection(set2))
# print("Intersection of set1, set2, and set3:", set1.intersection(set2, set3))

# Union
print("Union of set1 and set2:", set1.union(set2))
print("Union of set1, set2, and set3:", set1.union(set2, set3))

# Difference
print("Difference of set1 and set2 (elements in set1 but not in set2):", set1.difference(set2))
print("Difference of set2 and set1 (elements in set2 but not in set1):", set2.difference(set1))

# Symmetric Difference
print("Symmetric difference of set1 and set2 (elements in either set but not both):", set1.symmetric_difference(set2))
```

```python
# demonstrate adding, removing, and modifying dictionary elements
caffeine_content_mg = {
    "Mr. Goodbar chocolate": 122,
    "Red Bull": 33,
    "Monster Hitman Sniper energy drink": 122,
    "Lipton Brisk iced tea - lemon flavor": 2,
    "dark chocolate coated coffee beans": 869,
    "Regular drip or percolated coffee": 60,
    "Buzz Bites Chocolate Chews": 1639,
}
# Add a new key-value pair
caffeine_content_mg["Espresso"] = 64
print("Dictionary after adding 'Espresso':")
display(caffeine_content_mg)
```

```python
caffeine_content_mg["Espresso"] = 5555 # if the Key already exist, updated it
display(caffeine_content_mg)
```

```python

# Modify an existing value
caffeine_content_mg["Red Bull"] = 111
print("Dictionary after modifying 'Red Bull':")
display( caffeine_content_mg)

# Remove a key-value pair
del caffeine_content_mg["Mr. Goodbar chocolate"]
print("Dictionary after removing 'Mr. Goodbar chocolate':")
display(caffeine_content_mg)

# Another way to remove a key-value pair and get the value
popped_value = caffeine_content_mg.pop("Lipton Brisk iced tea - lemon flavor")
print("Popped value:", popped_value)
print("Dictionary after popping 'Lipton Brisk iced tea - lemon flavor':")
display(caffeine_content_mg)
```
