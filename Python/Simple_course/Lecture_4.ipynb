```python
# the focus is on the print vs display

import numpy as np
import pandas as pd

# Create a random matrix
matrix = np.random.rand(5, 5)

# Print the matrix
print("Printing the matrix:")
print(matrix)

# Display the matrix as a pandas DataFrame for better formatting
print("\nDisplaying the matrix:")
display(pd.DataFrame(matrix)) # colab (Jy Notebook specific)
```

```python
""" Computes the total cost of leasing a car given the down payment,
    monthly rate, and number of months """
#

down_payment = 7
payment_per_month = 10
num_months = 19

total_cost = down_payment + (payment_per_month * num_months)

print (f"Total cost: ${total_cost:.2f}")
```

```python
x = 10
w = 12

# Reassigning x
x = x + w
print(f"After reassignment (x = x + w): {x}")

# Using augmented assignment
x = 10 # Reset x for the next example
x += w
print(f"After augmented assignment (x += w): {x}")
```

```python
import math

# Calculating the square root
num = 49
num_sqrt = math.sqrt(num)
print(f"The square root of {num} is: {num_sqrt}")

# Using trigonometric functions (e.g., sine)
print(f"The sine of pi/2 is: {math.sin(math.pi/2)}")

# Calculating the distance between two points
x1 = 7
y1 = -3
x2 = -3
y2 = -9

x_dist_sq = math.pow((x2 - x1), 2) # Square of the difference in x-coordinates
y_dist_sq = math.pow((y2 - y1), 2) # Square of the difference in y-coordinates
points_distance = math.sqrt(x_dist_sq + y_dist_sq) # Square root of the sum of squares

print(f"The distance between the two points is: {points_distance:.4f}")
```
