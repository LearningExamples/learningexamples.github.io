```python
# Create two lists
list1 = [10, 5, 8, 12, 5]
list2 = [10, 3, 8, 15, 6]

print("Original list1:", list1)
print("Original list2:", list2)

# Check if the lists have the same length
if len(list1) != len(list2):
    print("Lists have different lengths, cannot compare element-wise.")
else:
    # Iterate through the lists and update values if the difference is zero
    for i in range(len(list1)):
        if list1[i] - list2[i] == 0:
            list1[i] = 0
            list2[i] = 0

    print("List1 after update:", list1)
    print("List2 after update:", list2)
```

```python
# Create two lists
list1 = [10, 5, 8, 12, 5]
list2 = [10, 3, 8, 15, 5]

print("Original list1:", list1)
print("Original list2:", list2)

# Check if the lists have the same length
if len(list1) != len(list2):
    print("Lists have different lengths, cannot compare element-wise.")
else:
    # Iterate through the lists and remove elements if the difference is zero
    # Iterate backwards to avoid index issues after popping
    for i in range(len(list1) - 1, -1, -1):
        if list1[i] - list2[i] == 0:
            list1.pop(i)
            list2.pop(i)

    print("List1 after removing elements:", list1)
    print("List2 after removing elements:", list2)
```

```python
# Example of sorting a list of strings by their length

string_list = ["apple", "banana", "cherry", "date"]
print("Original list:", string_list)

# Define a custom function to return the length of the string
def get_length(item):
  return len(item)

# Sort the list using the custom function as the key
string_list.sort(key=get_length)
print("Sorted list by length:", string_list)

# You can also use a lambda function directly
string_list = ["apple", "banana", "cherry", "date"] # Reset the list
string_list.sort(key=lambda item: len(item))
print("Sorted list by length using lambda:", string_list)
```

# Dictionary

```python
# Create a dictionary of students and their grades
student_grades = {
    "Alice": 95,
    "Bob": 88,
    "Charlie": 75,
    "David": 92
}

print("Original dictionary:", student_grades)
display(student_grades)
```

```python
# Delete a key-value pair
del student_grades["Bob"]

print("Dictionary after deleting 'Bob':", student_grades)
```

```python
# You can also use .pop() to remove a key and get its value
removed_grade = student_grades.pop("Charlie")
print("Dictionary after popping 'Charlie':", student_grades)
print("Removed grade:", removed_grade)
```

```python
# If you try to pop a key that doesn't exist, it will raise a KeyError
# You can provide a default value to avoid this error
removed_grade_safe = student_grades.pop("Eve", "Student not found")
print("Attempting to pop 'Eve':", removed_grade_safe)
```

```python
# Example using clear()
my_dict = {"Ahmad": 1, "Jane": 42}
print("Original dictionary for clear():", my_dict)
my_dict.clear()
print("Dictionary after clear():", my_dict)
```

```python
# Example using update()
my_dict = {"Ahmad": 1, "Jane": 42}
print("Original dictionary for update():", my_dict)
my_dict.update({"John": 50, "Ahmad": 2}) # Update existing key and add a new one
print("Dictionary after update():", my_dict)
```

```python
# Example using pop()
my_dict = {"Ahmad": 1, "Jane": 42, "Charlie": 75}
print("Original dictionary for pop():", my_dict)
# Pop an existing key
popped_value = my_dict.pop("Jane")
print("Dictionary after popping 'Jane':", my_dict)
print("Popped value:", popped_value)

# # Pop a non-existing key with a default value
popped_value_safe = my_dict.pop("Eve", "Key not found")
print("Attempting to pop 'Eve' with default:", popped_value_safe)
print("Dictionary after attempting to pop 'Eve':", my_dict)

# If you don't provide a default and the key doesn't exist, it raises a KeyError
# try:
#     my_dict.pop("Frank")
# except KeyError as e:
#     print(f"KeyError: {e}")
```

```python
# pop vs get ?
my_dict = {"Ahmad": 1, "Jane": 42, "Charlie": 75}
x = my_dict.pop("Jane")
print(my_dict) # Jane is deleted
print("value for Jane is", x)


# get
my_dict = {"Ahmad": 1, "Jane": 42, "Charlie": 75}
x = my_dict.get("Jane")
print(my_dict) # Jane is not deleted
print("value for Jane is", x)


# when pop without defult it gives an error if not found, in get it returns None
```

```python
# dict, we want to iterate it
my_dict = {"Ahmad": 1, "Jane": 42, "Charlie": 75}
for name, grade, in my_dict.items():
  print(f"Name: {name}, Grade: {grade}")
```

```python
# iterate with keys
my_dict = {"Ahmad": 1, "Jane": 42, "Charlie": 75}
for name in my_dict:
  print("the key is ", name)
  # value
  print("\t\tthe value is ",  my_dict[name]) # my_dict.get(name)
```

```python
# iterate with keys
my_dict = {"Ahmad": 1, "Jane": 42, "Charlie": 75}
print(f"when using {my_dict.items()= }\n") # kinda list

for k,v in my_dict.items():
  print("the key is ", k)
  print("\t\tthe value is ", v)
```

```python
# iterate with keys
my_dict = {"Ahmad": 1, "Jane": 42, "Charlie": 75}
print(f"when using {my_dict.keys() = }\n")


for name in my_dict.keys():
  print("the key is ", name)
  # value
  print("the value is ", my_dict.get(name, "not found")) # my_dict[name]
```

```python
print(f"when using {my_dict.values() = }")
for grade in my_dict.values():
  print("the value is ", grade)
```

```python
print(f"when using {my_dict.values() = }")
print(f"when using this is a list {list(my_dict.values()) = }")
```

```python
my_dict = {"apple": 1, "banana": 2, "cherry": 3}
print("Original dictionary:", my_dict)

# Convert keys to a list
keys_list = list(my_dict.keys())
print("List of keys:", keys_list)

# Convert values to a list
values_list = list(my_dict.values())
print("List of values:", values_list)

# Convert items (key-value pairs) to a list of tuples
items_list = list(my_dict.items())
print("List of items (tuples):", items_list)
```

# Nested Dict

```python
# Example of a nested dictionary
nested_dict = {
    "person1": {
        "name": "Alice",
        "age": 30,
        "city": "New York"
    },
    "person2": {
        "name": "Bob",
        "age": 25,
        "city": "London"
    },
    "person3": {
        "name": "Charlie",
        "age": 35,
        "city": "Paris"
    }
}

print("Original nested dictionary:", nested_dict)
```

```python
# Accessing elements in a nested dictionary
print("\nAccessing elements:")
print("Person 1's name:", nested_dict["person1"]["name"])
print("Person 2's age:", nested_dict["person2"]["age"])
```

```python

# Modifying elements in a nested dictionary
nested_dict["person1"]["age"] = 31
print("\nNested dictionary after modifying age:", nested_dict)

# Adding a new key-value pair to an inner dictionary
nested_dict["person2"]["occupation"] = "Engineer"
print("Nested dictionary after adding occupation:", nested_dict)
print("--"*10)
display(nested_dict)
```

```python
# Adding a new outer key-value pair
nested_dict["person4"] = {"name": "David", "age": 28, "city": "Tokyo"}
print("Nested dictionary after adding a new person:", nested_dict)
display(nested_dict)
print("--"*10)


# Adding a new outer key-value pair
nested_dict["person5"] = {"name": "David"}
print("Nested dictionary after adding a new person:", nested_dict)
display(nested_dict)
print("--"*10)

# Adding a new outer key-value pair
nested_dict["person6"] = "N/A"
print("Nested dictionary after adding a new person:", nested_dict)
display(nested_dict)
print("--"*10)



```

```python
nested_dict = {
    "person1": {
        "name": "Alice",
        "age": 30,
        "city": "New York"
    },
    "person2": {
        "name": "Bob",
        "age": 25,
        "city": "London"
    },
    "person3": {
        "name": "Charlie",
        "age": 35,
        "city": "Paris"
    }
}
# Iterating through a nested dictionary
print("\nIterating through the nested dictionary:")
for person, details in nested_dict.items():
    print(f"\n{person}:")
    for key, value in details.items():
        print(f"  {key}: {value}")
```
