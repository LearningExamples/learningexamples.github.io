```python
input_text = input("Enter a number: ")
float_variable = float(input_text)
int_variable = int(float_variable)

print(f"original input text: {input_text}")
print(f"input text converted to a float: {float_variable}")
print(f"float variable converted to an int: {int_variable}")
```

```python
# Binary to decimal conversion
binary_number = "1101"
decimal_number = int(binary_number, 2)
print(f"Binary: {binary_number}, Decimal: {decimal_number}")

# Decimal to binary conversion
decimal_to_convert = 13
binary_converted = bin(decimal_to_convert)[2:] # [2:] to remove the '0b' prefix
print(f"Decimal: {decimal_to_convert}, Binary: {binary_converted}")
```

```python
user_num = 7

div_remainder = user_num % 2
print("The remainder is :" , div_remainder)

if div_remainder == 0:
    print(f"{user_num} is even.")
else:
    print(f"{user_num} is odd.")
```

```python
a = (div_remainder == 1) # no need for () but it looks better :)
print(a)
```

```python
user_num = 7

div_remainder = user_num % 2
print("The remainder is :" , div_remainder)

cond = div_remainder == 0  # this will be either True or False
print("Cond is ", cond)
if cond:
    print(f"{user_num} is even.")
else:
    print(f"{user_num} is odd.")
```

```python
user_num = 7

div_remainder = user_num % 2 # 1

if div_remainder != 0:
    print(f"{user_num} is odd.")
else:
    print(f"{user_num} is even.")
```
