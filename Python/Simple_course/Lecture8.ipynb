```python

my_dict = {"A": 1, "B": 2, "C": 3}

# Check if key "B" exists
if "B" in my_dict:
   print("Found \"B\" as a key")
else:
   print("\"B\" not found as a key")

# Membership operator does not check values directly with 'in' on the dictionary
if 3 in my_dict:
    print("Found 3 directly in dict (checks keys): False") # This will print False
else:
    print("3 not found directly in dict (checks keys)")

# To check if a value exists, you need to check the dictionary's values view
if 3 in my_dict.values():
    print("Found 3 in dict values")
else:
    print("3 not found in dict values")

# To check if a key-value pair exists, you can check the items view
if ("C", 3) in my_dict.items():
    print("Found ('C', 3) in dict items")
else:
    print(" ('C', 3) not found in dict items")
```
