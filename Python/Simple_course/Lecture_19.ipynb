```python
import math as m
radius = 5

# We access functions inside the module using 'dot notation' (math.something)
area = m.pi * (radius ** 2)
root = m.sqrt(16)

print(f"The area is: {area:.2f}")
print(f"The square root of 16 is: {root}")
```


| Style | Syntax | Pros/Cons |
|---|---|---|
| Standard | `import math` | Best Practice. It is clear where the function comes from (e.g., math.sqrt). |
| Specific | `from math import sqrt` | You can just type sqrt(25). Good for short scripts, but can cause naming conflicts if you have your own function named sqrt. |
| Alias | `import numpy as np` | Renames the module to save typing. Standard in data science (e.g., np.array). |

The "Reloading" Trap (Crucial for Notebooks)

This is the number one issue students face in Colab.

If a student finds a bug in geometry_tools.py, fixes it, and re-runs the import cell, nothing will change. Python "remembers" (caches) the first import to save time.

You must teach them how to reload a module without restarting the whole runtime.

The Fix: If they modify the module file, they must use importlib in their main block to see the changes.

```python
import geometry_tools
import importlib
# from importlib import reload   # --> I can use reload()

# Force Python to re-read the file because we changed code inside it!
importlib.reload(geometry_tools)

# Calculate area of a 10x5 rectangle
rect = geometry_tools.rectangle_area(10, 5)
print(f"Rectangle Area: {rect}")

# Calculate area of a triangle
tri = geometry_tools.triangle_area(10, 5)
print(f"Triangle Area: {tri}")
```
