```python
hourly_rate = 20

while True:
    hours_input = input("Enter hours worked (or 'done' to quit): ")

    if hours_input.lower() == 'done' or hours_input.lower() == 'quit':
        print("Exiting salary calculator.")
        break

    is_valid_integer = False
    processed_hours = 0.0

    # Defensive check for integer input (handles negative integers)
    if hours_input.startswith('-'):
        # Check if the part after the '-' is all digits
        if hours_input[1:].isdigit():
            is_valid_integer = True
            processed_hours = float(hours_input) # Convert to float
    elif hours_input.isdigit():
        is_valid_integer = True
        processed_hours = float(hours_input) # Convert to float

    if is_valid_integer:
        salary = processed_hours * hourly_rate
        print(f"Calculated salary: ${salary:.2f}")
    else:
        print("Invalid input. Please enter a whole number for hours or 'done' to quit.")
```

| Type | Reason exception is raised |
| :--------- | :---------------------------------------------------------------------------------- |
| `EOFError` | `input()` hits an end-of-file condition (EOF) without reading any input. |
| `KeyError` | A dictionary key is not found in the set of keys. |
| `ZeroDivisionError` | Divide by zero error |
| `ValueError` | Invalid value (Ex: Input mismatch) |
| `IndexError` | Index is out of bounds. |

```python
while True:
    num1_input = 12
    num2_input = input("Enter the second number: (or 'done' to quit):")
    if num2_input.lower() == 'done' or num2_input.lower() == 'quit':
        print("Exiting calculation.")
        break


    try:
        num1 = float(num1_input)
        num2 = float(num2_input)
        result = num1 / num2
        print(f"Result: {result}")
    except ValueError:
        print("Invalid input. Please ensure both inputs are valid numbers.")
    except ZeroDivisionError:
        print("Error: Cannot divide by zero. Please enter a non-zero second number.")
    except Exception as e: # Catch any other unexpected errors
        print(f"An unexpected error occurred: {e}")
```

```python
hourly_rate = 20

while True:
    hours_input = input("Enter hours worked (or 'done' to quit): ")

    if hours_input.lower() == 'done' or hours_input.lower() == 'quit':
        print("Exiting salary calculator.")
        break

    try:
        hours = float(hours_input)

        # Raise an exception if hours are negative
        if hours < 0:
            raise ValueError("Hours worked cannot be negative.")

        salary = hours * hourly_rate
        print(f"Calculated salary: ${salary:.2f}")
    except ValueError as e: # Catch ValueError for both invalid string conversion and negative hours
        print(f"Invalid input: {e}. Please enter a non-negative number for hours or 'done' to quit.")
    except Exception as e: # Catch any other unexpected errors
        print(f"An unexpected error occurred: {e}")
```

```python
def calculate_salary_safe(hours, hourly_rate):
    if hours < 0:
        raise ValueError("in the function, Hours worked cannot be negative.")
    salary = hours * hourly_rate
    return salary


hourly_rate = 20

while True:
    hours_input = input("Enter hours worked (or 'done' to quit): ")

    if hours_input.lower() == 'done' or hours_input.lower() == 'quit':
        print("Exiting salary calculator.")
        break

    try:
        hours = float(hours_input)
        # Call the function that might raise a ValueError
        calculated_salary = calculate_salary_safe(hours, hourly_rate)
        print(f"Calculated salary: ${calculated_salary:.2f}")
    except ValueError as e:
        print(f"Invalid input: {e}. Please enter a non-negative number for hours or 'done' to quit.")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")
```

```python
print("--- Starting division attempt ---")

num1 = 10
num2_input = input("Enter a number to divide by (or 'error' to force a TypeError): ")

try:
    # Attempt to convert input to a number and perform division
    if num2_input.lower() == 'error':
        # Force a TypeError for demonstration
        result = num1 / 'invalid_type'
    else:
        num2 = float(num2_input)
        result = num1 / num2
        print(f"Division successful. Result: {result}")
except ZeroDivisionError:
    print("Error: Cannot divide by zero!")
except ValueError:
    print("Error: Invalid input. Please enter a valid number.")
except TypeError:
    print("Error: Detected invalid type for division.")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
finally:
    print("--- Division attempt finished ---")
```

```python
print("Attempting division...")

try:
    # This block is where you put code that might cause an error
    numerator = 10
    denominator = int(input("Enter a number to divide by: "))
    result = numerator / denominator
    print(f"Result of division: {result}")
except ZeroDivisionError:
    # This block runs if a ZeroDivisionError occurs in the 'try' block
    print("Error: You cannot divide by zero!")
except ValueError:
    # This block runs if a ValueError occurs (e.g., non-integer input)
    print("Error: Invalid input. Please enter a whole number.")
finally:
    # This block always runs, regardless of whether an error occurred or not
    print("Division attempt concluded.")

```

```python
def divide_numbers(a, b):
    try:
        result = a / b
        print(f"Division successful. Result: {result}")
        return result
    except ZeroDivisionError:
        print("Error: Cannot divide by zero!")
        return None
    except TypeError:
        print("Error: Both inputs must be numbers!")
        return None
    finally:
        print("--- Division attempt finished ---")


# Test cases
print("Attempt 1: Valid division")
divide_numbers(10, 2)

print("\nAttempt 2: Division by zero")
divide_numbers(10, 0)

print("\nAttempt 3: Invalid input type")
divide_numbers(10, "hello")

print("\nAttempt 4: Another valid division")
divide_numbers(15, 3)
```
