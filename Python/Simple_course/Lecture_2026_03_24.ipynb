````python
print("--- Starting division attempt ---")

try:
    num1 = 10
    num2_input = input("Enter a number to divide by (or 'error' to force a TypeError): ")
    num2 = float(num2_input) # string --> error
    result = num1 / num2 #  10/0 --> error
    print(f"The result of {num1} divided by {num2} is: {result:.2f}")

except ZeroDivisionError:
    print("Error: Cannot divide by zero!")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
finally:
    print("--- Division attempt finished ---")
````

````python
def divide_numbers(a, b):
    try:
        result = a / b
        print(f"Division successful. Result: {result}")
        return result
    except ZeroDivisionError:
        print("Error: Cannot divide by zero!")
        return None
    except TypeError:
        print("Error: Both inputs must be numbers!")
        return None
    finally:
        print("--- Division attempt finished ---")


  # Test cases
print("Attempt 1: Valid division")
divide_numbers(10, 2)

print("\nAttempt 2: Division by zero")
divide_numbers(10, 0)

print("\nAttempt 3: Invalid input type")
divide_numbers(10, "hello")

print("\nAttempt 4: Another valid division")
divide_numbers(15, 3)
````

The "Reloading" Trap (Crucial for Notebooks)

This is the number one issue students face in Colab.

If a student finds a bug in geometry_tools.py, fixes it, and re-runs the import cell, nothing will change. Python "remembers" (caches) the first import to save time.

You must teach them how to reload a module without restarting the whole runtime.

The Fix: If they modify the module file, they must use importlib in their main block to see the changes.

````python
import geometry_tools
import importlib
# from importlib import reload   # --> I can use reload()

# Force Python to re-read the file because we changed code inside it!
importlib.reload(geometry_tools)

# Calculate area of a 10x5 rectangle
rect = geometry_tools.rectangle_area(10, 5)
print(f"Rectangle Area: {rect}")

# Calculate area of a triangle
tri = geometry_tools.triangle_area(10, 5)
print(f"Triangle Area: {tri}")
````
