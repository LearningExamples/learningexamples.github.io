# Factorial Operation

```python
# Factorial calculation in a while loop
num = 7  # Change this number to calculate the factorial of a different number
factorial = 1
i = 1

while i <= num:
    factorial *= i
    i += 1

print(f"The factorial of {num} is {factorial}")
```

```python
n = 7
ans = 1
while n > 0:
  ans = ans * n
  n = n - 1
print("the factorial is ", ans)
```

# Nested Loops

```python
for i in range(3):
  print("i = ", i)
  for j in range(4):
    print(f"\t{i = }, {j = }")
  print("----------------------------")
print("I am finally out of the loopS")
```

```python
my_car_list = ["Toyota", "Honda", "Ford"]
my_discount = [10,20]
for car in my_car_list:
  print("starting the loop 1:", car)
  for discount in my_discount:
    print("\tcar: ", car, "discount: ", discount)
  print("----------------------------finished with car: ", car)
```

```python
# Example of a for loop iterating through a dictionary
my_dict = {"a": 1, "b": 2, "c": 3}

print("Iterating through keys:")
for key in my_dict:
  print(key)


print("access the value using key and loops")
for key in my_dict:
  print("key= ", key," value= ", my_dict[key])

print("\nIterating through values:")
for value in my_dict.values():
  print(value)

print("\nIterating through key-value pairs:")
for key, value in my_dict.items():
  print(f"Key: {key}, Value: {value}")
```

```python
# Examples of using range with for loops

# Example 1: range(stop) - iterates from 0 up to (but not including) stop
print("Example 1: range(5)")
for i in range(5):
  print(i)
```

```python
print("\nExample 2: range(start, stop) - iterates from start up to (but not including) stop")
# Example 2: range(start, stop) - iterates from start up to (but not including) stop
print("Example 2: range(2, 7)")
for i in range(2, 7):
  print(i)
```

```python



print("\nExample 3: range(start, stop, step) - iterates from start up to (but not including) stop with a specified step")
# Example 3: range(start, stop, step) - iterates from start up to (but not including) stop with a specified step
print("Example 3: range(0, 10, 2)")
for i in range(0, 10, 2):
  print(i)

print("\nExample 4: Using range with a negative step")
# Example 4: Using range with a negative step
print("Example 4: range(10, 0, -1)")
for i in range(10, 0, -1):
  print(i)
```
