```python
# write code that creates 10 random number and prints them in loop
import random
count = 0
random.seed(1238)  # the random number are always the same when run
while count < 10:
  print("my number is :", random.randint(1, 2 * count + 5))
  print("Count: ", count, "the upper bound is: ", 2 * count + 5)
  count += 1

print("Count: ", count, "the upper bound is: ", 2 * count + 5)
```

```python


# while loop with a break statement (simulating do-while)
print("\nSimulating a do-while loop:")
i = 0
while True:
    print(f"i is: {i}")
    i += 1
    if i >= 5:
        break

print("\nDo-while simulation finished.")

# Another way to simulate do-while
print("\nAnother do-while simulation:")
j = 0
# The condition is checked after the first iteration
while True:
    print(f"j is: {j}")
    j += 1
    if j > 5:
        break
```

```python
# factorial calculator
n = 5
sol = 1

while n >= 1:
  sol = n * sol
  n = n-1

print(f"{sol}")
```

```python
# factorial calculator
n = 5

sol = 1

while n >= 1:
  if n != 1:
    print(f"{n} ", end ="* ")
  else:
    print(f"{n} ", end ="=")
  sol = n * sol
  n = n-1

print(f"{sol}")
```

```python
names = [
    "Biffle",
    "Bowyer",
    "Busch",
    "Gordon",
    "Patrick"
]

for item in names:
    print(item, "|", end=" ")

print("\nPrinting in reverse:")
for name in reversed(names):
    print(name, "|", end=" ")
```
