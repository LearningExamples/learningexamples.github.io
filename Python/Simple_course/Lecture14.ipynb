### Slicing a string

```python
# Example of list slicing
my_list = ['a','b','c', 'd', 'e', 'f', 'g', 'h', 'i', 'j'] # if my list was list of char, it would do the same
print("Original list: ", my_list)
sliced_list = my_list[2:7]  # Slice from index 2 up to (but not including) index 7
print("sliced list: " ,sliced_list)
```

```python
# Example with a step
sliced_list_with_step = my_list[1:9:2] # Slice from index 1 to 9 with a step of 2
print(sliced_list_with_step)
```

```python
# Example with negative indices
negative_slice = my_list[-5:-2] # Slice using negative indices
print(negative_slice)
```

## string Methods

```python
# Example using replace()
original_string = "Hello world! I am new to world, but not WORLD"
new_string = original_string.replace("world", "Python") # this is case senstive
print(f"Original string: {original_string}")
print(f"String after replace(): {new_string}")
```

```python
# # Example replacing all occurrences
another_string = "This is a test test test."
replaced_string = another_string.replace("test", "example")
print(f"Original string: {another_string}")
print(f"String after replacing all 'test': {replaced_string}")
```

```python
# Example replacing a specific number of occurrences
limited_replace = another_string.replace("test", "sample", 2)
print(f"Original string: {another_string}")
print(f"String after replacing the first 2 'test': {limited_replace}")
```

```python
my_string = "programming is fun and programming is useful"

# find(x)
print(f"Index of first 'is': {my_string.find('is')}")
print(f"Index of 'python' (not found): {my_string.find('python')}")
```

```python
# find(x, start)
print(f"Index of 'programming' starting from index 15: {my_string.find('programming', 15)}")

# find(x, start, end)
print(f"Index of 'programming' between index 15 and 30: {my_string.find('programming', 15, 30)}") # on the location 30 we break the programming word into smaller
```

```python
# rfind(x)
print(f"Index of last 'programming': {my_string.rfind('programming')}")
```

## Methods to create new strings:



    capitalize() -- Returns a copy of the string with the first character capitalized and the rest lowercased.
    lower() -- Returns a copy of the string with all characters lowercased.
    upper() -- Returns a copy of the string with all characters uppercased.
    strip() -- Returns a copy of the string with leading and trailing whitespace removed.
    title() -- Returns a copy of the string as a title, with first letters of words capitalized.


```python
my_str = "Hello bob, WHAT are you doing HERE!!"
print(my_str)
x = my_str.capitalize()
print(x)
```

```python
my_str = "Hello bob, WHAT are you doing HERE!!"
print(my_str)
x = my_str.title()
print(x)
```

```python
my_str = "              Hello bob, WHAT are you doing HERE!!"
print(my_str)
x = my_str.strip()
print(x)
```

```python
# combine them
my_str = "              Hello bob, WHAT are you doing HERE!!"
print(my_str)
x = my_str.strip().title()
print(x)
```

```python
# join a list into string
my_tokens = ["bob", "Alex","A","montreal"]
seperator = "_"
my_info = seperator.join(my_tokens)
print(my_info)

my_info = "**".join(my_tokens)
print(my_info)

#old way was with loop and +
my_new_str = ""
for token in my_tokens:
    my_new_str += token + "_"
print(my_new_str)
```
