```python
import math

def calc_pizza_area(pizza_diameter):
  pizza_radius = pizza_diameter / 2.0
  pizza_area = math.pi * pizza_radius * pizza_radius
  return pizza_area

print(f"12.0 inch pizza is {calc_pizza_area(12.0):.3f} square inches")
print(f"16.0 inch pizza is {calc_pizza_area(16.0):.3f} square inches")
```

```python
def my_step_function(x):
  y = 0
  if x <= -5:
    y = -5
  elif x > -5 and x <= 0:
    y = x
  elif x > 0 and x < 3:
    y = 0
  elif x >=3 and x <= 5:
    y = -x
  else:
    y = -5

  return y

for x in range(-10, 11):
  print("x:", x, "\tf(x):", my_step_function(x))

```

```python
# beyond class
# plot the function
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-6, 6, 100)
y = np.array([my_step_function(val) for val in x])
plt.plot(x, y)
plt.show()
```

```python
# create function x and y and return x^y using loops
def my_pow(x,y):
  my_result = 1
  for i in range(0, y):
    my_result = my_result * x
  return my_result

print(my_pow(7,3))
```

```python
# the function has access to these:
CM_PER_INCH = 2.54
INCHES_PER_FOOT = 12

def height_US_to_cm(feet, inches):
  """Converts height in feet/inches to centimeters
  feet , inches ,......."""
  total_inches = feet * INCHES_PER_FOOT + inches
  cm = total_inches * CM_PER_INCH
  return cm

feet = 6
inches = 4
centimeters = height_US_to_cm(feet, inches)
print(f"Centimeters: {centimeters}")
print("CM_PER_INCH:", CM_PER_INCH)
print("INCHES_PER_FOOT:", INCHES_PER_FOOT)
```
